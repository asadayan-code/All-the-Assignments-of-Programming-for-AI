```python
def calculate_results(marks_list):
  total = sum(marks_list)
  percentage = (total / 300) * 100
  if percentage >= 80:
    grade = "A"
  elif percentage >= 70:
    grade = "B"
  elif percentage >= 60:
    grade = "C"
  elif percentage >= 50:
    grade = "D"
  else:
    grade = "F"
  return total, percentage, grade


def main():
  students = []
  for i in range(5):
    print(f"\nEnter details for student {i+1}:")
    name = input("Enter name: ")
    roll_no = input("Enter roll number: ")

    print("Enter marks for 3 subjects (out of 100 each):")
    python_marks = float(input("Python: "))
    ai_marks = float(input("Artificial Intelligence: "))
    math_marks = float(input("Mathematics: "))

    marks_list = [python_marks, ai_marks, math_marks]
    total, percentage, grade = calculate_results(marks_list)

    student_dict = {
        "name": name,
        "roll_no": roll_no,
        "marks": {
            "Python": python_marks,
            "AI": ai_marks,
            "Mathematics": math_marks,
        },
        "total": total,
        "percentage": percentage,
        "grade": grade,
    }
    students.append(student_dict)

  print("\n--- Student Results ---")
  highest_percentage = -1
  top_student = ""

  with open("results.txt", "w") as file:
    file.write("--- Student Results ---\n")
    for s in students:
      res_str = (
          f"Roll No: {s['roll_no']}, Name: {s['name']}, Total Marks:"
          f" {s['total']}/300, Percentage: {s['percentage']:.2f}%, Grade:"
          f" {s['grade']}\n"
      )
      print(res_str.strip())
      file.write(res_str)

      if s["percentage"] > highest_percentage:
        highest_percentage = s["percentage"]
        top_student = s["name"]

    top_str = (
        f"\nTop Student: {top_student} with Percentage:"
        f" {highest_percentage:.2f}%\n"
    )
    print(top_str)
    file.write(top_str)


if __name__ == "__main__":
  main()
```

```python
def main():
  sentence = input("\nEnter a sentence: ")

  words = sentence.split()
  total_words = len(words)
  total_chars = len(sentence)

  longest_word = max(words, key=len) if words else ""
  shortest_word = min(words, key=len) if words else ""

  vowels = "aeiouAEIOU"
  vowel_count = sum(1 for char in sentence if char in vowels)

  unique_words = set(words)
  sorted_words = sorted(words)

  analysis_result = f"""--- Sentence Analysis ---
Sentence: {sentence}
Total Words: {total_words}
Total Characters: {total_chars}
Longest Word: {longest_word}
Shortest Word: {shortest_word}
Vowel Count: {vowel_count}
Unique Words: {unique_words}
Words in Alphabetical Order: {sorted_words}
"""

  print(analysis_result)

  with open("word_analysis.txt", "w") as file:
    file.write(analysis_result)


if __name__ == "__main__":
  main()
```

```python
def calculate_bill(products):
  complete_bill = 0
  most_expensive = {"name": "", "price": -1}

  for p in products:
    p["total_price"] = p["price"] * p["quantity"]
    complete_bill += p["total_price"]

    if p["price"] > most_expensive["price"]:
      most_expensive = {"name": p["name"], "price": p["price"]}

  discount = 0
  if complete_bill > 10000:
    discount = complete_bill * 0.10

  final_bill = complete_bill - discount
  return complete_bill, discount, final_bill, most_expensive


def main():
  products = []
  print("\nEnter details for 5 products:")
  for i in range(5):
    print(f"\nProduct {i+1}:")
    name = input("Product name: ")
    price = float(input("Price: "))
    quantity = int(input("Quantity: "))
    products.append({"name": name, "price": price, "quantity": quantity})

  complete_bill, discount, final_bill, most_expensive = calculate_bill(products)

  output = "--- Shopping Bill ---\n"
  for p in products:
    output += (
        f"{p['name']} - Price: Rs. {p['price']} | Qty: {p['quantity']} | Total:"
        f" Rs. {p['total_price']}\n"
    )

  output += f"\nComplete Bill: Rs. {complete_bill:.2f}\n"
  if discount > 0:
    output += f"Discount (10%): -Rs. {discount:.2f}\n"
  output += f"Final Bill: Rs. {final_bill:.2f}\n"
  output += (
      f"Most Expensive Product: {most_expensive['name']} (Rs."
      f" {most_expensive['price']})\n"
  )

  print(output)

  with open("bill.txt", "w") as file:
    file.write(output)


if __name__ == "__main__":
  main()
```

```python
def add(a, b):
  return a + b


def subtract(a, b):
  return a - b


def multiply(a, b):
  return a * b


def divide(a, b):
  if b == 0:
    return "Error! Division by zero."
  return a / b


def modulus(a, b):
  if b == 0:
    return "Error! Division by zero."
  return a % b


def main():
  while True:
    print("\n--- Calculator ---")
    try:
      num1 = float(input("Enter first number: "))
      num2 = float(input("Enter second number: "))
    except ValueError:
      print("Invalid input. Please enter numbers.")
      continue

    print("Select an operation:")
    print("1. Addition (+)")
    print("2. Subtraction (-)")
    print("3. Multiplication (*)")
    print("4. Division (/)")
    print("5. Modulus (%)")

    choice = input("Enter choice (+, -, *, /, %): ")

    if choice == "+":
      result = add(num1, num2)
    elif choice == "-":
      result = subtract(num1, num2)
    elif choice == "*":
      result = multiply(num1, num2)
    elif choice == "/":
      result = divide(num1, num2)
    elif choice == "%":
      result = modulus(num1, num2)
    else:
      print("Invalid operation selected.")
      continue

    print(f"Result: {result}")

    cont = input("Do you want to perform another calculation? (yes/no): ")
    if cont.lower() != "yes":
      print("Exiting calculator. Goodbye!")
      break


if __name__ == "__main__":
  main()
```

```python
def main():
  menu = {
      1: {"name": "Burger", "price": 500},
      2: {"name": "Pizza", "price": 800},
      3: {"name": "Fries", "price": 250},
      4: {"name": "Drink", "price": 150},
  }

  print("\n--- Restaurant Menu ---")
  for key, item in menu.items():
    print(f"{key}. {item['name']} - Rs. {item['price']}")

  order_list = []

  while True:
    try:
      choice = int(
          input("\nEnter item number to order (1-4) or 0 to finish order: ")
      )
    except ValueError:
      print("Invalid input. Please enter a number.")
      continue

    if choice == 0:
      break
    elif choice in menu:
      qty = int(input(f"Enter quantity for {menu[choice]['name']}: "))
      item_total = menu[choice]["price"] * qty
      order_list.append({
          "name": menu[choice]["name"],
          "price": menu[choice]["price"],
          "quantity": qty,
          "total": item_total,
      })
      print(f"Added {qty} x {menu[choice]['name']} to your order.")
    else:
      print("Invalid menu choice. Try again.")

  if order_list:
    final_bill = sum(item["total"] for item in order_list)
    bill_receipt = "--- Restaurant Bill / Orders ---\n"
    for item in order_list:
      bill_receipt += (
          f"{item['name']} x {item['quantity']} @ Rs. {item['price']} = Rs."
          f" {item['total']}\n"
      )
    bill_receipt += f"\nTotal Final Bill: Rs. {final_bill}\n"

    print("\n" + bill_receipt)

    with open("orders.txt", "w") as file:
      file.write(bill_receipt)
    print("Bill saved to orders.txt successfully!")
  else:
    print("No items ordered.")


if __name__ == "__main__":
  main()
```
